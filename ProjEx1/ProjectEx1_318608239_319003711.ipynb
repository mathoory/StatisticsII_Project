## Dataset description
This project uses the US Accidents (2016–2023) dataset from Kaggle and focuses exclusively on incidents where the **reported City is New York**.

the US Accidents (2016–2023) dataset is a nationwide car accident dataset covering 49 U.S. states, the data come from several live traffic APIs that stream incident reports and their properties. In total, the dataset includes about 7.7 million accident records.

From the full dataset, we extract only a small set of traffic-impact–related variables and compute additional features.
All rows containing missing or invalid values in these selected fields are removed to ensure a clean and reliable subset for analysis.

## Column selection

For this analysis we retain a focused subset of fields from the original dataset, standardized to consistent names. All rows containing missing or invalid values in any of these fields were removed, ensuring a clean dataset for analysis. The final columns are:

1. **accident_id**  
   - **Type:** categorical (identifier)  
   - **Description:** Numeric accident identifier extracted from the original `ID` field by removing the `A-` prefix and retaining only digits.  
   - **Stored as:** `int64`

2. **severity_of_traffic**  
   - **Type:** categorical (ordinal, 1–4)  
   - **Description:** Severity score indicating the traffic impact of the accident, where higher values represent greater disruption.  
   - **Stored as:** `int8`

3. **traffic_disruption_duration**  
   - **Type:** continuous  
   - **Description:** Duration of the accident’s effect on traffic flow, computed as `End_Time − Start_Time` and expressed in hours.  
   - **Stored as:** `float64`

4. **distance_of_traffic**  
   - **Type:** continuous  
   - **Description:** Distance in miles over which normal traffic flow was affected, taken from the original `Distance(mi)` field.  
   - **Stored as:** `float64`

5. **is_day**  
   - **Type:** binary  
   - **Description:** Indicates whether the accident occurred during daylight (`1 = Day`, `0 = Night`), derived from `Sunrise_Sunset`.  
   - **Stored as:** `int8`

6. **is_traffic_signal**  
   - **Type:** binary  
   - **Description:** Indicates the presence of a traffic signal near the accident location (`1 = Yes`, `0 = No`).  
   - **Stored as:** `int8`

## Data Transformation & Manipulation
- Filter the dataset to include only records where `City = "New York"` (case-insensitive).
- Convert `Start_Time` and `End_Time` to datetime objects and compute **traffic_disruption_duration** in hours.
- Generate **is_day** from `Sunrise_Sunset` (`1 = Day`, `0 = Night`) and drop the original field.
- Convert `Distance(mi)` and `traffic_disruption_duration` to numeric types, marking invalid or negative values as null.
- Extract the numeric portion of `ID` to form **accident_id** and standardize all column names to snake_case.
- Remove all rows containing any null, invalid, or non-computable values across the selected fields.
- Retain only the final cleaned columns:
\
`accident_id`, `severity_of_traffic`, `traffic_disruption_duration`, `distance_of_traffic`, `is_day`, `is_traffic_signal`.
- Export the processed New York subset to `new_york_accidents_prepared.csv`.


## Data Preparation (Previously Completed)
The raw Kaggle dataset (2.84GB) has been processed, filtered for New York accidents, cleaned, and transformed into its final structure as explained above. 

The prepared dataset is hosted in a Google Sheets document. Thus, there is no need to run this preparation step again. and it is only here to show the pre-process step.


```python
import os
import pandas as pd
import numpy as np

# -------- SETTINGS -------- #
force_recreate = False   # set to True if you want to regenerate the dataset
# -------------------------- #

csv_path = r"C:\Users\Mattan\Downloads\US_Accidents_March23.csv"
output_dir = os.path.abspath(".")
clean_csv_path = os.path.join(output_dir, "new_york_accidents_prepared.csv")

usecols = [
    "ID", "Severity", "Start_Time", "End_Time", "Distance(mi)",
    "City", "Sunrise_Sunset", "Traffic_Signal"
]

# Skip processing if file exists and recreation is not forced
if os.path.exists(clean_csv_path) and not force_recreate:
    pass  
else:
    df_raw = pd.read_csv(csv_path, usecols=usecols)

    # Filter for New York
    df_raw["City"] = df_raw["City"].astype(str).str.strip()
    df_ny = df_raw[df_raw["City"].str.casefold() == "new york"].copy()

    # Datetime conversion + duration
    df_ny["Start_Time"] = pd.to_datetime(df_ny["Start_Time"], errors="coerce")
    df_ny["End_Time"] = pd.to_datetime(df_ny["End_Time"], errors="coerce")
    df_ny["traffic_disruption_duration"] = (
        (df_ny["End_Time"] - df_ny["Start_Time"]).dt.total_seconds() / 3600
    )

    # Day/Night
    df_ny["Sunrise_Sunset"] = df_ny["Sunrise_Sunset"].astype(str).str.strip().str.title()
    df_ny["is_day"] = (df_ny["Sunrise_Sunset"] == "Day").astype("int8")

    # Traffic Signal
    df_ny["is_traffic_signal"] = df_ny["Traffic_Signal"].astype("int8")

    # Distance
    df_ny["Distance(mi)"] = pd.to_numeric(df_ny["Distance(mi)"], errors="coerce")
    df_ny["distance_of_traffic"] = df_ny["Distance(mi)"]

    # Severity
    df_ny["severity_of_traffic"] = pd.to_numeric(df_ny["Severity"], errors="coerce")

    # Clean ID → only digits → accident_id
    df_ny["accident_id"] = (
        df_ny["ID"].astype(str).str.extract(r"(\d+)", expand=False)
    )
    df_ny["accident_id"] = pd.to_numeric(df_ny["accident_id"], errors="coerce")

    # Keep final cleaned columns
    df = df_ny[
        [
            "accident_id",
            "severity_of_traffic",
            "traffic_disruption_duration",
            "distance_of_traffic",
            "is_day",
            "is_traffic_signal",
        ]
    ].copy()

    # Invalid → null
    df.loc[df["traffic_disruption_duration"] < 0, "traffic_disruption_duration"] = np.nan
    df.loc[df["distance_of_traffic"] < 0, "distance_of_traffic"] = np.nan

    # Drop rows with ANY nulls
    df = df.dropna().reset_index(drop=True)

    # Final types
    df["accident_id"] = df["accident_id"].astype("int64")
    df["severity_of_traffic"] = df["severity_of_traffic"].astype("int8")
    df["traffic_disruption_duration"] = df["traffic_disruption_duration"].astype("float64")
    df["distance_of_traffic"] = df["distance_of_traffic"].astype("float64")
    df["is_day"] = df["is_day"].astype("int8")
    df["is_traffic_signal"] = df["is_traffic_signal"].astype("int8")

    # Save once
    df.to_csv(clean_csv_path, index=False)
```

## Binary Varaibles

```python
import matplotlib.pyplot as plt

# Binary Distributions
day_counts = df['is_day'].value_counts().rename(index={0: 'Night', 1: 'Day'})
day_perc = (day_counts / day_counts.sum() * 100).round(2)

# Traffic signal distribution with percentages
signal_counts = df['is_traffic_signal'].value_counts().rename(index={0: 'No Signal', 1: 'Signal'})
signal_perc = (signal_counts / signal_counts.sum() * 100).round(2)

# Side-by-side bar plots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Day vs Night (left)
colors_day = ['#1f77b4', '#ff7f0e']  # Day, Night
axes[0].bar(day_counts.index, day_counts.values, color=colors_day[:len(day_counts)], edgecolor='black')
axes[0].set_title('Accidents time (New York)', fontsize=12)
axes[0].set_xlabel('Time of day')
axes[0].set_ylabel('Count')
axes[0].set_xticks(range(len(day_counts)))
axes[0].set_xticklabels(day_counts.index, rotation=0)

# Expand y-limit so annotations fit inside plot
ymax_day = day_counts.max()
axes[0].set_ylim(0, ymax_day * 1.12)

# Annotate counts + percentages
for i, label in enumerate(day_counts.index):
    cnt = int(day_counts.iloc[i])
    pct = day_perc.iloc[i]
    axes[0].text(i, cnt + ymax_day * 0.02, f"{cnt}\n({pct}%)", ha='center', va='bottom', fontsize=9, clip_on=False)

# Traffic Signal (right)
colors_sig = ['#6c757d', '#2ca02c']  # No Signal, Signal
axes[1].bar(signal_counts.index, signal_counts.values, color=colors_sig[:len(signal_counts)], edgecolor='black')
axes[1].set_title('Traffic Signal Presence (New York)', fontsize=12)
axes[1].set_xlabel('Traffic Signal')
axes[1].set_ylabel('Count')
axes[1].set_xticks(range(len(signal_counts)))
axes[1].set_xticklabels(signal_counts.index, rotation=0)

# Expand y-limit so annotations fit inside plot
ymax_sig = signal_counts.max()
axes[1].set_ylim(0, ymax_sig * 1.12)

# Annotate counts + percentages
for i, label in enumerate(signal_counts.index):
    cnt = int(signal_counts.iloc[i])
    pct = signal_perc.iloc[i]
    axes[1].text(i, cnt + ymax_sig * 0.02, f"{cnt}\n({pct}%)", ha='center', va='bottom', fontsize=9, clip_on=False)

plt.tight_layout()
plt.show()
```

The first plot shows that most accidents in New York occur during the day: 12,717 cases (67.35%) compared to 6,166 cases (32.65%) at night.
The second plot shows the distribution of accidents by traffic-signal presence. Most accidents (14,764 cases, 78.19%) occur in areas without traffic signals, while 4,119 accidents (21.81%) occur at signalized locations.

This result does not imply causation and could be influenced by various factors such as traffic volume during the day, and road design.

## Continous

```python
from IPython.display import display
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Numerical summaries
num_cols = ['traffic_disruption_duration','distance_of_traffic']
desc = df[num_cols].describe(percentiles=[0.25,0.5,0.75, 0.99]).T

display(
    desc.style.format("{:,.4f}").set_caption("Numerical summary").background_gradient(cmap="Blues")
)

# Outliers
top_data = {col: df[col].nlargest(15).values for col in num_cols}
df_top10 = pd.DataFrame(top_data).T
df_top10.columns = [f'#{i+1}' for i in range(15)]
display(
    df_top10.style
    .format("{:,.4f}")
    .set_caption("Top Highest Values (Outlier Inspection)")
    .background_gradient(cmap="Reds", axis=1) # 'Reds' highlights the extreme highs better
)

def plot_clipped_hist(col, color, display_name, unit):
    data = df[col].dropna()
    p99 = np.percentile(data, 99)

    fig, axs = plt.subplots(1, 2, figsize=(16, 5))

    # Left: clipped at 99th percentile
    sns.histplot(data[data <= p99], bins=35, color=color, ax=axs[0])
    axs[0].set_title(f'{display_name} Bottom 99% (<= {p99:.2f} {unit})')
    axs[0].set_xlabel(f'{display_name} ({unit})')
    axs[0].set_ylabel('Count')
    axs[0].set_xlim(0, p99)
    axs[0].set_yscale('log')

    # Right: unclipped
    sns.histplot(data, bins=75, color=color, ax=axs[1])
    axs[1].set_title(f'{display_name} Full distribution')
    axs[1].set_xlabel(f'{display_name} ({unit})')
    axs[1].set_ylabel('Count')
    axs[1].set_yscale('log')

    plt.tight_layout()
    plt.show()

# Run the function for both numeric columns with readable names and units
colors = ["#FF7F0E", "#1F77B4"] # orange, blue

plot_clipped_hist(
    'traffic_disruption_duration',
    colors[0],
    'Traffic Disruption Duration',
    'hours'
)

plot_clipped_hist(
    'distance_of_traffic',
    colors[1],
    'Distance of Traffic Affected',
    'miles'
)
```

Both `traffic_disruption_duration` and `distance_of_traffic` show extremely right-skewed distributions with a large concentration of very small values and a small number of extreme outliers.

* The maximum `traffic_disruption_duration` is 3943, while 99% of values are below 15.9, and the median is only 1.26.
* `distance_of_traffic` has a median of 0.137, but a maximum of 8.5, and 25%-50% of observations are exactly zero.

These characteristics cause the raw histograms to appear compressed on the left side, with a tall spike near zero and almost no visible distribution across the remaining range. The combination of heavy-tailed outliers and many zero values makes a standard histogram unsuitable without transformation.

To better visualize the data, we use log-scaled histograms and histograms clipped at the 99th percentile to provide a clearer representation of the typical range of values while still acknowledging the presence of rare, extreme events.

## Data visualization

Boxplots representing the dist. of traffic disruption duration compared according to severity of traffic impact

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filtered version for the boxen plot
df_filtered = df[df['severity_of_traffic'] != 1]

fig, axes = plt.subplots(1, 2, figsize=(18, 8), sharey=True)

# -------------------------
# LEFT PLOT – Boxplot (all severities)
# -------------------------
sns.boxplot(
    data=df, 
    x='severity_of_traffic', 
    y='traffic_disruption_duration',
    ax=axes[0]
)
axes[0].set_title('Traffic Disruption Duration by Severity (Boxplot)', fontsize=14)
axes[0].set_xlabel('Severity', fontsize=12)
axes[0].set_ylabel('Duration (hours)', fontsize=12)
axes[0].set_yscale('log')
axes[0].grid(True, which="both", ls="--", alpha=0.3)

# -------------------------
# RIGHT PLOT – Boxen Plot (severity != 1)
# -------------------------
sns.boxenplot(
    data=df_filtered,
    x='severity_of_traffic',
    y='traffic_disruption_duration',
    k_depth='trustworthy',
    ax=axes[1]
)
axes[1].set_title('Duration by Severity (Boxen Plot, Severity ≠ 1)', fontsize=14)
axes[1].set_xlabel('Severity', fontsize=12)
axes[1].set_ylabel('')  # remove duplicate y-label
axes[1].set_yscale('log')
axes[1].grid(True, which="both", ls="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

We used a Letter-Value (Boxen) plot to overcome the many outliers common when analyzing large, heavy-tailed datasets by recursively estimating quantiles. 


All categories show similar median disruption durations, indicating that typical accidents are brief regardless of severity. Severity 4 exhibits a distinct "fat tail," demonstrating that while typical accidents are brief across all classes, the probability of extreme disruption is higher in this category.

# Research Questions


**A. Regression Question (continuous -> continuous)**  
Do accidents that affect a longer distance of traffic lead to longer traffic‐disruption durations? \
e.g.: Does an increase in `distance_of_traffic` lead to increase in `traffic_disruption_duration`?

**B. Regression Question (continuous -> binary)**  
Does an increase in the distance of traffic affected increase the probability that the accident occurred at night? \
e.g.: Does an increase in **distance_of_traffic** affect the probability that `is_day = 0`?

**C. Hypothesis Test (continuous difference between binary categories)**  
Is the traffic_disruption_duration different between accidents that occur with a nearby traffic signal versus without one? \
e.g.: Is `traffic_disruption_duration` different between groups defined by `is_traffic_signal = 1` vs. `is_traffic_signal = 0`?

Note on question B: Although the assignment requires framing the binary variable (`is_day`) as the outcome in a logistic model, we note that it cannot be causally influenced by the predictor (`distance_of_traffic`). Daytime is fixed independently of accident characteristics.
The model should not be interpreted causally: it simply evaluates whether the distance affected is associated with the time of day.
We considered creating a derived binary variable (e.g., impact > average) that would make the causal direction more natural, but chose to keep the original variables structure.
